## Filter out matches pre 2006, use 4 slams all 9 atp 1000,
## Clay: RG, MC, MA, RM, BAR, MUN, HAM, RIO
## Hard: AO, USO, PAR, IW, MIA, CIN, TOR/MONT, SHA
## Grass: WIM, HALLE, LON, HERTOGENBOSCH, MALLORCA, EASTBOURNE

```python
import pandas as pd
from pathlib import Path

RAW_DIRECTORY = Path("../data/raw/sackmann")

tour_keep = ["Wimbledon", "Roland Garros", "US Open", "Australian Open", "Madrid", "Montecarlo",
          "Rome", "Barcelona", "Munich", "Hamburg", "Rio de Janeiro", "Shanghai", "Paris", "Indian Wells",
          "Cincinnati", "Miami", "Toronto", "Montreal", "Halle", "Queen's Club", "Monte Carlo Masters", "Paris Masters",
          "Cincinnati Masters", "Indian Wells Master", "Miami Masters", "Rome Masters", "Madrid Masters",
          "Canada Masters", "Shangai Masters", "s Hertogenbosch", "Mallorca", "Eastbourne"]

files = sorted(RAW_DIRECTORY.glob("atp_matches_*.csv"))
matches = pd.concat((pd.read_csv(f) for f in files), ignore_index=True)
```

```python
matches["tourney_date"] = pd.to_datetime(matches["tourney_date"], format="%Y%m%d")
matches = matches[matches["tourney_date"] >= "2006-01-01"]
matches = matches.sort_values("tourney_date", ascending = True)
matches.head()
```

```python
matches = matches[matches["tourney_name"].isin(tour_keep)]
matches.head()
```

```python
matches.to_csv("../data/processed/atp_matches.csv", index = False)   
```
